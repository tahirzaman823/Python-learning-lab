```python
# Import Libraries
import pandas as pd

dataset = pd.read_csv("Customer Call List.csv")
dataset.head()

```

```python
# Column 1 First_Name    strip whitespace + standardize capitalization
dataset['First_Name'] = dataset['First_Name'].str.strip().str.title()
# dataset['First_Name']

# Column 2 Clean the columns — strip whitespace(Unwanted like / _ . ) + standardize capitalization 
dataset['Last_Name'] = dataset['Last_Name'].str.strip('/._').str.title()
# dataset['Last_Name']

# Colum 3  Remove everything except digits
dataset['Phone_Number'] = dataset['Phone_Number'].str.replace('[^0-9]', '', regex=True)
# dataset['Phone_Number'] 

# Column 4 Correct The Formate Y,n  Yes,No
dataset['Paying Customer'] = dataset['Paying Customer'].replace({'Y': 'Yes', 'N': 'No'})
# dataset['Paying Customer']

# Column 5 Correct The Formate Y,n  Yes,No
dataset['Do_Not_Contact'] = dataset['Do_Not_Contact'].replace({'Y': 'Yes', 'N': 'Yes'}) 
# dataset['Do_Not_Contact']

# Column 6,7,8 are useless
dataset =  dataset.drop(columns=['Not_Useful_Column', 'Unnamed: 8', 'Unnamed: 9'])
```

```python
# Save the cleaned dataset
dataset.to_csv("cleaned_data.csv", index=False)
```
